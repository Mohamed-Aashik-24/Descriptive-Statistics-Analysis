```python
import pandas as pd
import matplotlib.pyplot as plt

print("Name: Mohamed Aashik S \nRoll No: 24BAD072\n")
```

###1) Histogram: Distribution of Happiness Score



```python
# Histogram for Happiness Score
plt.figure(figsize=(6,4))
plt.hist(df['HappinessScore'], bins=20)
plt.xlabel('Happiness Score')
plt.ylabel('Frequency')
plt.title('Distribution of Happiness Score')
plt.show()
```

###2) Boxplot: Happiness Score


```python
# Boxplot
plt.figure(figsize=(6,4))
plt.boxplot(df['HappinessScore'].dropna())
plt.ylabel('Happiness Score')
plt.title('Boxplot of Happiness Score')
plt.show()
```

### Statistical Summary for Happiness Score (Histogram & Boxplot):
*   **Count**: 142
*   **Mean**: 5.37
*   **Min**: 1.11, **Max**: 7.54

###3) Scatter Plot: GDP vs. Happiness Score


```python
# Scatter Plot: GDP vs Happiness Score
plt.figure(figsize=(6,4))
plt.scatter(df['Log GDP per capita'], df['HappinessScore'])
plt.xlabel('GDP per Capita')
plt.ylabel('Happiness Score')
plt.title('GDP vs Happiness Score')
plt.show()
```

### Statistical Summary for GDP vs. Happiness Score (Scatter Plot):
*   **HappinessScore**: Mean ~5.37, Range 1.11-7.54
*   **Log GDP per capita**: Count 133, Mean ~9.35, Range 6.63-11.47

###4) Bar Chart: Top 10 Happiest Countries


```python
# Top 10 Happiest Countries
top10 = df.sort_values('HappinessScore', ascending=False).head(10)
plt.figure(figsize=(8,4))
plt.bar(top10['Country'], top10['HappinessScore'])
plt.xticks(rotation=45, ha='right')
plt.xlabel('Country')
plt.ylabel('Happiness Score')
plt.title('Top 10 Happiest Countries')
plt.show()
```

### Statistical Summary for Top 10 Happiest Countries (Bar Chart):
*   **Top 10 HappinessScore**: Count 10, Mean ~7.40, Range 7.21-7.54 (very narrow spread)

###5) Correlation Matrix Plot


```python
# Correlation Matrix Plot
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Correlation Matrix')
plt.show()
```

### Statistical Summary for Correlation Matrix:
*   **Correlation Coefficients**: Values between -1 and +1, indicating strength and direction of linear relationships between all numerical variables.

###6) Line Plot: Happiness Score by Rank


```python
plt.figure(figsize=(10,6))
plt.plot(df['Rank'], df['HappinessScore'])
plt.xlabel('Rank')
plt.ylabel('Happiness Score')
plt.title('Happiness Score by Rank')
plt.grid(True)
plt.show()
```

### Statistical Summary for Happiness Score by Rank (Line Plot):
*   **HappinessScore**: Mean ~5.37, Range 1.11-7.54
*   **Rank**: Count 140, Mean ~70.5, Range 1-140 (ordered sequence)
